### Part 1 — Minimize $f(x) = x^2 - 4x + 4$

**Critical point.** Set the first derivative to zero:

$$f'(x) = 2x - 4 = 0 \;\Longrightarrow\; x = 2$$

**Classify.** Use the second derivative test:

$$f''(x) = 2 > 0 \quad \text{for all } x$$

Since $f'' > 0$, the function is convex everywhere and $x = 2$ is a **minimum**.

**Minimum value.**

$$f(2) = (2)^2 - 4(2) + 4 = 4 - 8 + 4 = 0$$

This is also visible by completing the square: $f(x) = (x-2)^2$, which is zero at
$x = 2$ and positive everywhere else. Because the function is convex, this local
minimum is also the **global** minimum.

---

### Part 2 — Optimize $f(x) = x^3 - 3x^2$

**Critical points.** Set the first derivative to zero and factor:

$$f'(x) = 3x^2 - 6x = 3x(x - 2) = 0 \;\Longrightarrow\; x = 0 \text{ and } x = 2$$

**Classify.** The second derivative is $f''(x) = 6x - 6$:

| $x$ | $f''(x)$ | Sign | Classification |
|---|---|---|---|
| $0$ | $6(0) - 6 = -6$ | negative | **local maximum** |
| $2$ | $6(2) - 6 = +6$ | positive | **local minimum** |

**Extreme values.**

$$f(0) = 0^3 - 3(0)^2 = 0 \qquad \text{(local maximum value)}$$
$$f(2) = 2^3 - 3(2)^2 = 8 - 12 = -4 \qquad \text{(local minimum value)}$$

**Summary:** a local maximum of $0$ at $x = 0$, and a local minimum of $-4$ at
$x = 2$.

Note these are **local** extrema only. A cubic is unbounded in both directions:
as $x \to \infty$, $f \to \infty$, and as $x \to -\infty$, $f \to -\infty$. The
function therefore has no global maximum or minimum.

### Part 3 — $f(x,y) = x^2 - y^2$

**Gradient.**

$$\nabla f = \left\langle \frac{\partial f}{\partial x},\; \frac{\partial f}{\partial y} \right\rangle = \langle 2x,\; -2y \rangle$$

**Critical point.** Set both components to zero:

$$2x = 0 \;\Rightarrow\; x = 0 \qquad -2y = 0 \;\Rightarrow\; y = 0$$

The only critical point is the origin $(0, 0)$, where $f(0,0) = 0$.

**Classify using the Hessian.**

$$H = \begin{bmatrix} f_{xx} & f_{xy} \\ f_{yx} & f_{yy} \end{bmatrix} = \begin{bmatrix} 2 & 0 \\ 0 & -2 \end{bmatrix}$$

$$\det(H) = (2)(-2) - (0)(0) = -4 < 0$$

A negative determinant means the critical point is a **saddle point**.

**Why.** The two cross-sections through the origin behave oppositely:

- Along the $x$-axis ($y = 0$): $f(x, 0) = x^2$, which has a **minimum** at the origin
- Along the $y$-axis ($x = 0$): $f(0, y) = -y^2$, which has a **maximum** at the origin

The point is a minimum in one direction and a maximum in a perpendicular
direction, so it is neither a maximum nor a minimum overall. The eigenvalues of
$H$ are $+2$ and $-2$ — opposite signs, which is the general signature of a
saddle.

## Q3:

Minimize $f(x,y,z) = x^2 + y^2 + z^2$ subject to $x + 2y + z = 6$.

### 1. Set up the Lagrangian

Write the constraint in the form $g(x,y,z) = 0$:

$$g(x,y,z) = x + 2y + z - 6 = 0$$

The Lagrangian introduces a multiplier $\lambda$ for the constraint:

$$\mathcal{L}(x,y,z,\lambda) = f - \lambda g = x^2 + y^2 + z^2 - \lambda\,(x + 2y + z - 6)$$

### 2. Derive the system of equations

Set every partial derivative of $\mathcal{L}$ to zero:

$$\frac{\partial \mathcal{L}}{\partial x} = 2x - \lambda = 0 \;\Longrightarrow\; x = \frac{\lambda}{2}$$
$$\frac{\partial \mathcal{L}}{\partial y} = 2y - 2\lambda = 0 \;\Longrightarrow\; y = \lambda$$
$$\frac{\partial \mathcal{L}}{\partial z} = 2z - \lambda = 0 \;\Longrightarrow\; z = \frac{\lambda}{2}$$
$$\frac{\partial \mathcal{L}}{\partial \lambda} = -(x + 2y + z - 6) = 0 \;\Longrightarrow\; x + 2y + z = 6$$

The first three equations are equivalent to the vector statement
$\nabla f = \lambda \nabla g$, and the fourth simply restores the original
constraint.

### 3. Solve for critical points

Substitute the first three expressions into the constraint:

$$\frac{\lambda}{2} + 2(\lambda) + \frac{\lambda}{2} = 6$$
$$\frac{\lambda}{2} + \frac{\lambda}{2} + 2\lambda = 6$$
$$3\lambda = 6 \;\Longrightarrow\; \boxed{\lambda = 2}$$

Back-substituting:

$$x = \frac{2}{2} = 1, \qquad y = 2, \qquad z = \frac{2}{2} = 1$$

**Critical point: $(x, y, z) = (1, 2, 1)$.**

Verify the constraint: $1 + 2(2) + 1 = 6$ ✓

Value of the objective:

$$f(1,2,1) = 1^2 + 2^2 + 1^2 = 6$$

Verify the gradient condition: $\nabla f = \langle 2, 4, 2\rangle$ and
$\nabla g = \langle 1, 2, 1\rangle$, so $\nabla f = 2\nabla g = \lambda \nabla g$ ✓

### 4. Classify the extremum

This critical point is a **minimum**, with minimum value $f = 6$.

Three arguments:

**Convexity.** $f = x^2 + y^2 + z^2$ is a convex function and the constraint set
is a plane, which is a convex set. A convex function minimized over a convex set
has a unique global minimum, and any critical point found is that minimum.

**Unbounded above.** Moving far from the origin in any direction *within* the
plane makes $f$ arbitrarily large, so no maximum exists on the constraint set.
The only extremum is the minimum.

**Geometric check.** $f$ is the squared distance from the origin, so the problem
asks for the point on the plane closest to the origin. The distance from the
origin to the plane $x + 2y + z = 6$ is

$$d = \frac{|{-6}|}{\sqrt{1^2 + 2^2 + 1^2}} = \frac{6}{\sqrt{6}} = \sqrt{6} \approx 2.449$$

and $d^2 = 6$, matching $f(1,2,1) = 6$ exactly.

### 5. The role of the constraint

**Without the constraint**, minimizing $x^2 + y^2 + z^2$ is trivial: the minimum
is $0$ at the origin. The constraint $x + 2y + z = 6$ **excludes the origin**,
since $0 + 0 + 0 = 0 \neq 6$. It confines the search to a two-dimensional plane
inside three-dimensional space, and the best achievable value rises from $0$ to
$6$.

**Geometrically**, the level surfaces of $f$ are spheres centred at the origin.
Expanding a sphere outward from the origin, the first moment it touches the plane
is the solution — and at that moment the sphere is *tangent* to the plane. That
tangency is exactly what $\nabla f = \lambda \nabla g$ expresses: the gradients
are parallel, so there is no direction along the plane that decreases $f$ further.

**The multiplier has a meaning.** $\lambda$ is the *shadow price* of the
constraint — the rate at which the optimal value changes as the constraint is
relaxed. For the general constraint $x + 2y + z = c$, the optimal value is
$f^*(c) = c^2/6$, so

$$\frac{df^*}{dc} = \frac{2c}{6} = \frac{c}{3}, \qquad \text{and at } c = 6: \quad \frac{df^*}{dc} = 2 = \lambda$$

So $\lambda = 2$ says that increasing the right-hand side from $6$ to $7$ raises
the minimum objective value by approximately $2$ units. (Checked numerically:
$f^*(6) = 6.000$ and $f^*(7) = 8.167$, a change of $2.167$, consistent with a
local rate of $2$ and the upward curvature of $f^*(c)$.)

## Q4:

$$J(w_1) = (w_1 - 5)^4 + (w_1 - 3)^2 + 4, \qquad \eta = 0.1, \qquad w_1^{(0)} = 2.0$$

### The gradient and the update rule

$$\frac{dJ}{dw_1} = 4(w_1 - 5)^3 + 2(w_1 - 3)$$

$$w_1^{(k+1)} = w_1^{(k)} - \eta \,\frac{dJ}{dw_1}\bigg|_{w_1^{(k)}}$$

### Five iterations

**Iteration 1** ($w_1^{(0)} = 2$):
$$\frac{dJ}{dw_1} = 4(2-5)^3 + 2(2-3) = 4(-27) + 2(-1) = -108 - 2 = -110$$
$$w_1^{(1)} = 2 - 0.1(-110) = 2 + 11 = \mathbf{13}$$

**Iteration 2** ($w_1^{(1)} = 13$):
$$\frac{dJ}{dw_1} = 4(8)^3 + 2(10) = 2048 + 20 = 2068$$
$$w_1^{(2)} = 13 - 0.1(2068) = 13 - 206.8 = \mathbf{-193.8}$$

**Iteration 3** ($w_1^{(2)} = -193.8$):
$$\frac{dJ}{dw_1} = 4(-198.8)^3 + 2(-196.8) = -3.14278 \times 10^7$$
$$w_1^{(3)} = -193.8 + 3.14278\times10^6 = \mathbf{3.14259 \times 10^6}$$

**Iteration 4:**
$$\frac{dJ}{dw_1} = 1.24143 \times 10^{20} \;\Longrightarrow\; w_1^{(4)} = \mathbf{-1.24143 \times 10^{19}}$$

**Iteration 5:**
$$\frac{dJ}{dw_1} = -7.65286 \times 10^{57} \;\Longrightarrow\; w_1^{(5)} = \mathbf{7.65286 \times 10^{56}}$$

| $k$ | $w_1^{(k)}$ | $dJ/dw_1$ | $w_1^{(k+1)}$ |
|---|---|---|---|
| 0 | $2$ | $-110$ | $13$ |
| 1 | $13$ | $2{,}068$ | $-193.8$ |
| 2 | $-193.8$ | $-3.143\times10^{7}$ | $3.143\times10^{6}$ |
| 3 | $3.143\times10^{6}$ | $1.241\times10^{20}$ | $-1.241\times10^{19}$ |
| 4 | $-1.241\times10^{19}$ | $-7.653\times10^{57}$ | $7.653\times10^{56}$ |

**The iteration diverges.** Each step overshoots the minimum and lands further
away than it started, with the magnitude growing and the sign alternating.

### Comparison with partial differentiation

Setting the derivative to zero directly:

$$4(w_1-5)^3 + 2(w_1-3) = 0$$

Substituting $u = w_1 - 5$ (so $w_1 - 3 = u + 2$):

$$4u^3 + 2(u+2) = 0 \;\Longrightarrow\; 2u^3 + u + 2 = 0$$

Solving this cubic gives $u \approx -0.835122$, so

$$\boxed{w_1^* \approx 4.1649}, \qquad J(w_1^*) \approx 5.8433$$

This is a minimum, and in fact the unique global minimum, because

$$\frac{d^2J}{dw_1^2} = 12(w_1-5)^2 + 2 \geq 2 > 0 \quad \text{for all } w_1$$

so $J$ is strictly convex.

### Why gradient descent failed here

Gradient descent is stable only when the step size satisfies
$\eta < 2/J''$ near the current iterate. At the starting point $w_1 = 2$:

$$J''(2) = 12(2-5)^2 + 2 = 110 \;\Longrightarrow\; \eta < \frac{2}{110} \approx 0.0182$$

The assigned $\eta = 0.1$ is about **5.5 times larger than the stability
limit**, so every step overshoots. Because the function is *quartic*, the
gradient grows as the cube of the distance from $w_1 = 5$, so each overshoot
produces a far larger gradient, and the divergence compounds explosively.

Repeating the same procedure with $\eta = 0.01$ converges to $w_1 = 4.164878$,
matching the analytical answer exactly.

| Method | Result |
|---|---|
| Partial differentiation | $w_1^* = 4.1649$ exactly, in one step |
| Gradient descent, $\eta = 0.1$ | Diverges to $7.65\times10^{56}$ after 5 iterations |
| Gradient descent, $\eta = 0.01$ | Converges to $4.164878$ |



## Q5:

$$\hat{w} = \arg\min_{w} \left\{ \|y - Xw\|^2 + \lambda\|w\|^2 \right\}$$

### Write the objective in matrix form

Using $\|v\|^2 = v^\top v$:

$$L(w) = (y - Xw)^\top(y - Xw) + \lambda\, w^\top w$$

### Expand the first term

$$(y - Xw)^\top(y - Xw) = y^\top y - y^\top Xw - w^\top X^\top y + w^\top X^\top Xw$$

The two middle terms are scalars and are transposes of one another, so
$y^\top Xw = (y^\top Xw)^\top = w^\top X^\top y$. They combine:

$$L(w) = y^\top y - 2w^\top X^\top y + w^\top X^\top Xw + \lambda\, w^\top w$$

### Take the gradient with respect to w

Using the standard matrix-calculus identities $\nabla_w(a^\top w) = a$ and
$\nabla_w(w^\top A w) = 2Aw$ for symmetric $A$ (note $X^\top X$ is symmetric):

| Term | Gradient |
|---|---|
| $y^\top y$ | $0$ (no $w$) |
| $-2w^\top X^\top y$ | $-2X^\top y$ |
| $w^\top X^\top Xw$ | $2X^\top Xw$ |
| $\lambda w^\top w$ | $2\lambda w$ |

$$\nabla_w L = -2X^\top y + 2X^\top Xw + 2\lambda w$$

### Set to zero and solve

$$-2X^\top y + 2X^\top Xw + 2\lambda w = 0$$

Divide by $2$ and collect the $w$ terms:

$$X^\top Xw + \lambda w = X^\top y$$
$$(X^\top X + \lambda I)\,w = X^\top y$$

$$\boxed{\hat{w}_{\text{ridge}} = \left(X^\top X + \lambda I\right)^{-1} X^\top y}$$

### How this changes the original solution

The ordinary least squares solution is

$$\hat{w}_{\text{OLS}} = (X^\top X)^{-1} X^\top y$$

The **only** change is the addition of $\lambda I$ to $X^\top X$ before
inverting — a constant $\lambda$ added along the diagonal. This is why the
method is also called *ridge* regression: a ridge is added down the diagonal of
the matrix.

Setting $\lambda = 0$ recovers OLS exactly.

### Impact of the penalty

**1. It guarantees invertibility.** $X^\top X$ is singular whenever the columns
of $X$ are linearly dependent, or whenever there are more features than
observations ($p > n$) — in which case OLS has no unique solution at all. For
any $\lambda > 0$, $X^\top X + \lambda I$ is strictly positive definite and
therefore always invertible. Ridge produces an answer where OLS cannot.

**2. It shrinks the weights toward zero.** Writing the SVD $X = U\Sigma V^\top$
with singular values $d_i$, the ridge solution scales each component by

$$\frac{d_i^2}{d_i^2 + \lambda}$$

This factor is always between $0$ and $1$, so every coefficient shrinks.
Directions with large $d_i$ (high-variance, well-determined directions) are
barely affected, while directions with small $d_i$ (poorly-determined, the ones
responsible for unstable estimates) are shrunk heavily. The penalty targets
exactly the unstable part of the solution.

**3. It trades bias for variance.** Ridge is a *biased* estimator —
$E[\hat{w}_{\text{ridge}}] \neq w_{\text{true}}$ for $\lambda > 0$ — but it has
substantially lower variance. For some $\lambda > 0$ the total mean squared
error is lower than OLS, which is why the method is used despite the bias.

**4. It controls overfitting.** Large weights let a model react sharply to small
input changes. Penalizing $\|w\|^2$ makes the fitted function smoother and
improves generalisation to unseen data.

**5. It does not produce sparsity.** L2 shrinks coefficients toward zero but
never exactly to zero. Zeroing out coefficients requires the L1 (Lasso) penalty
$\lambda\|w\|_1$ instead.

**6. The limits.** As $\lambda \to 0$, $\hat{w}_{\text{ridge}} \to \hat{w}_{\text{OLS}}$.
As $\lambda \to \infty$, $\hat{w}_{\text{ridge}} \to 0$. The parameter $\lambda$
interpolates between fitting the data and keeping the weights small, and is
normally chosen by cross-validation.

In practice the intercept term is left unpenalized, since shrinking it would
make the fit depend on where the origin of $y$ happens to sit.

## Q6:

The simple linear regression model is

$$y_i = w_0 + w_1 x_i + \varepsilon_i, \qquad i = 1, \dots, n$$

where $\varepsilon_i$ is the random error term. The following assumptions are made.

### 1. Linearity

The relationship between the predictor and the response is **linear in the
parameters**. The expected value of $y$ given $x$ is

$$E[y_i \mid x_i] = w_0 + w_1 x_i$$

This concerns linearity in $w_0$ and $w_1$, not in $x$. A model such as
$y = w_0 + w_1 x^2$ is still "linear regression" because it is linear in the
weights.

*If violated:* the fitted line systematically misses the true pattern, residuals
show curvature, and predictions are biased.

### 2. Independence of errors

The errors are mutually independent, so

$$\text{Cov}(\varepsilon_i, \varepsilon_j) = 0 \quad \text{for all } i \neq j$$

Knowing one observation's error tells you nothing about another's.

*If violated:* common with time-series data, where consecutive errors are
correlated (autocorrelation). The coefficient estimates remain unbiased but
standard errors are wrong, so confidence intervals and $p$-values cannot be
trusted.

### 3. Normality of errors

The errors are normally distributed:

$$\varepsilon_i \sim N(0, \sigma^2)$$

This assumption is **not required** to compute the least-squares estimates
themselves. It is required for the inference built on top of them — $t$-tests,
confidence intervals on $\hat{w}_0$ and $\hat{w}_1$, and prediction intervals.

*If violated:* point estimates are still valid, but the confidence intervals are
unreliable, particularly for small $n$.

### 4. Equal variance of errors (homoscedasticity)

The error variance is constant across all values of $x$:

$$\text{Var}(\varepsilon_i) = \sigma^2 \quad \text{for all } i$$

The spread of the points about the line should be the same at every $x$.

*If violated (heteroscedasticity):* the estimates remain unbiased, but OLS is no
longer the minimum-variance estimator and the standard errors are wrong. A
residual plot that fans out or funnels in is the usual symptom.

### 5. Zero mean error

$$E[\varepsilon_i] = 0 \quad \text{for all } i$$

Errors are random scatter about the line, not a systematic offset. This is what
makes $\hat{w}_0$ and $\hat{w}_1$ unbiased estimators of the true parameters.

### 6. The predictor is fixed and measured without error

The values $x_i$ are treated as known constants rather than random variables,
and are assumed free of measurement error.

*If violated:* measurement error in $x$ biases the slope estimate toward zero,
an effect known as regression dilution or attenuation bias.

### 7. The predictor varies

$$\sum_{i}(x_i - \bar{x})^2 > 0$$

The $x$ values must not all be identical. If they were, the denominator of

$$\hat{w}_1 = \frac{\sum(x_i - \bar{x})(y_i - \bar{y})}{\sum(x_i - \bar{x})^2}$$

would be zero and the slope would be undefined. Intuitively, a line cannot be
fitted through a single vertical stack of points.

### Summary

The four classical assumptions are often remembered as **LINE**:

| Letter | Assumption |
|---|---|
| **L** | **L**inearity |
| **I** | **I**ndependence of errors |
| **N** | **N**ormality of errors |
| **E** | **E**qual variance (homoscedasticity) |

taken together with zero-mean errors, a fixed and error-free predictor, and
non-constant $x$.

Compactly, assumptions 2 through 5 can be written as a single statement:

$$\varepsilon_i \stackrel{\text{iid}}{\sim} N(0, \sigma^2)$$

```python
import torch

# Schwarzschild radius and radial coordinate
rs_value = 0.00887
r_value  = 30000.1

# float64 matters here — see the note below
r  = torch.tensor(r_value, dtype=torch.float64, requires_grad=True)
rs = torch.tensor(rs_value, dtype=torch.float64)

# tau = 2r * sqrt(r/rs) * sqrt(1 - rs/r)
tau = 2 * r * torch.sqrt(r / rs) * torch.sqrt(1 - rs / r)

# reverse-mode autodiff populates r.grad with dtau/dr
tau.backward()

print(f"r      = {r_value}")
print(f"rs     = {rs_value}")
print(f"tau    = {tau.item():.6f}")
print(f"dtau/dr = {r.grad.item():.6f}")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

url = 'https://raw.githubusercontent.com/rahulbhadani/Datasets/master/RAV4/2020-04-15-05-12-00_2T3Y1RFV8KC014025_CAN_Messages_5_Hz.csv'
df = pd.read_csv(url)

X = df[['steer_angle']]
y = df['yaw_rate']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25)

model = LinearRegression().fit(X_train, y_train)

print("Slope:     ", model.coef_[0])
print("Intercept: ", model.intercept_)
print("R^2 (test):", model.score(X_test, y_test))
```
